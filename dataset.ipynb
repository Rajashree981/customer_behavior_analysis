```python
import pandas as pd
```

```python
df=pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
#check summary statistics of the dataset by default of numerical columns only
df.describe(include='all')
#check summary statistics of the dataset by including all columns, both numerical and categorical
```

```python
df.isnull().sum()
```

```python
#finding median of clothing and footwearcategory review score and fill missing data with median value
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))

```

```python
df.isnull().sum()
```

```python
#fixing column names by replacing spaces with underscores and converting to lowercase (snake_case)
df.columns=df.columns.str.replace(' ','_').str.lower()
```

```python
df.rename(columns={ 'purchase_amount_(usd)':'purchase_amount'},inplace=True)
df.columns
```

```python
#new column to group customers into age groups
labels=['young_adult','adult','middle_ages','senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
#qcut is used to create equal-sized bins based on the distribution of the 'age' column, and labels are assigned to each bin to create the new 'age_group' column.
```

```python
df[['age','age_group']].head(10)
```

```python
#purchasefrequency_days ,textual frequencies into no. of days after which the customer made a purchase again
frequency_mapping={'Fortnightly':14,'Weekly':7,'Monthly':30,'Quarterly':90,'Bi-Weekly':14,'Annually':365,'Every 3 Months':90}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
#map function is used to replace the textual frequencies in the 'purchase_frequency' column with their corresponding numerical values based on the defined mapping, and the new 'purchase_frequency_days' column is created to store these numerical values.
```

```python
df[['discount_applied','promo_code_used']].head(10)
#do we actually need these columns? 
```

```python
(df['discount_applied']==df['promo_code_used']).all()
#checking if the 'discount_applied' and 'promo_code_used' columns have the same
```

```python
#since both same so remove promo_code_used column
df.drop('promo_code_used',axis=1,inplace=True)
#inplace=True is used to modify the original DataFrame directly without creating a new one, and axis=1 specifies that we want to drop a column (as opposed to a row).
```
